# Laborator 4. NLP pentru text

Prelucrarea limbajului natural scris presupune procesarea, interpretarea și înțelegerea textului de către calculator.

Cele mai comune aplicații ale procesării textului sunt corectarea automată, analiza sentimentelor sau a emoțiilor, rezumarea automată, detectarea limbii și traducerile automate.

Sistemele bazate pe procesarea textului includ 1.preprocesarea, 2.reprezentarea, 3.analiza și modelare și 4.evaluarea.

În etape de **preprocesare** se realizează, determinarea frazelor, propozițiilor, curățarea textului (eliminarea semnelor de punctuație și a spațiilor multiple), transformarea în litere mici, tokenizarea (împărțirea textului în cuvinte sau subcuvinte), eliminarea cuvintelor de legătură ("stopwords"), identificarea părților de vorbire, identificarea numelor de entități și datelor calendaristice, lematizare (reducerea cuvintelor la forma de bază).

În etapa de **reprezentare**, se extrag carateristicile textului brut (șir de caractere). Cu alte cuvinte, textul este convertit în vectori de trăsături numerice pe care algoritmii de învățare îi pot modela. În general, reprezentările numerice ale textului pot fi de trei tipuri: simple (BoW, TF-IDF), bazate pe semantică (Word2Vec, GloVe, FastText) și contextuale (BERT, RoBERTa, GPT).

Etapa de **analiză și modelare** presupune învățarea datelor pe baza setului de trăsături obținut în etapa anterioară. Practic, prin învățare se dorește identificarea modelelor/regulilor ce permit diferențierea informațiilor existente, folosind algoritmi statistici sau neuronali. Algoritmii de învățare populari în NLP pentru text sunt: algoritmi clasici (ML) - Logistic Regression, Naive Bayes, SVM, Random Forest; rețetele neuronale simple: RNN, LSTM, GRU, CNN; și modele avansate: BERT, DisrilBERT, RoBERTa, GPT.

Ultima etapă, cea de **evaluare** presupune realizarea măsurătorilor pentru a determina performanțele sistemelor. Principalele mărimi folosite pentru evaluare sunt acuratețea, precizia, recall, scorul F1 și perplexitatea pentru generarea textului.

Creați o secvență de cod care preprocesează textul stocat în variabila text. Textul va fi la alegere dar va conține nume de entități și diverse semne de punctuație.

2. Analiza sentimentelor

Completați funcția *preprocess* din secvența de cod următoare astfel încât acesta să realizeze analiza sentimentelor pentru textul din variabila *text*


```python
import string

positive_lexicon = {
    "bun", "excelent", "frumos", "minunat", "plăcut", "mulțumit",
    "grozav", "superb", "fericit", "uimitor", "pozitiv"
}

negative_lexicon = {
    "rău", "urât", "oribil", "teribil", "slab", "dezamăgitor",
    "groaznic", "jalnic", "supărat", "negativ"
}

def preprocess(text):
    # litere mici
    text = text.lower()
    # eliminare semne de punctuație
    text = text.translate(str.maketrans("", "", string.punctuation))
    # tokenizare simplă
    tokens = text.split()
    return tokens

def sentiment_analysis_ro(text):
    tokens = preprocess(text)
    score = 0

    for token in tokens:
        if token in positive_lexicon:
            score += 1
        elif token in negative_lexicon:
            score -= 1

    if score > 0:
        return "sentiment pozitiv", score
    elif score < 0:
        return "sentiment negativ", score
    else:
        return "sentiment neutru", score


# Exemplu de utilizare:
text = "Filmul a fost foarte bun și frumos, dar finalul a fost cam dezamăgitor."
result = sentiment_analysis_ro(text)
print(result)
```

**Observație:** în text apar „bun” și „frumos” (+2) și „dezamăgitor” (−1), deci scorul este 1 și sentimentul este pozitiv.

```python

import string
import spacy
nlp = spacy.load("ro_core_news_sm")

positive_lexicon = {
    "bun", "excelent", "frumos", "minunat", "plăcut", "mulțumit",
    "grozav", "superb", "fericit", "uimitor", "pozitiv"
}

negative_lexicon = {
    "rău", "urât", "oribil", "teribil", "slab", "dezamăgitor",
    "groaznic", "jalnic", "supărat", "negativ"
}

def preprocess(text):
    # litere mici
    text = text.lower()
    # eliminare semne de punctuație
    text = text.translate(str.maketrans("", "", string.punctuation))
    # tokenizare cu lematizare
    tokens = nlp(text)
    lemas = []
    for token in tokens:
        lemas.append(token.lemma_)
    return lemas

def sentiment_analysis_ro(text):
    tokens = preprocess(text)
    score = 0

    for token in tokens:
        if token in positive_lexicon:
            score += 1
        elif token in negative_lexicon:
            score -= 1

    if score > 0:
        return "sentiment pozitiv", score
    elif score < 0:
        return "sentiment negativ", score
    else:
        return "sentiment neutru", score


# Exemplu de utilizare:
text = "Experiența a fost foarte bună și frumoasă, dar finalul a fost cam dezamăgitor."
result = sentiment_analysis_ro(text)
print(result)

```

**Observație:** acum „bună” devine „bun” și „frumoasă” devine „frumos”, deci sunt recunoscute în lexicon. Scorul este 2 − 1 = 1, sentiment pozitiv, ceea ce corespunde sensului real al propoziției.
