<h2 id="1"> 1. Introducere(Operatii pe vecinatate) </h2>
<p> Laboratorul trecut, ați aflat despre îmbunătățirea imaginilor prin modificarea individuală a fiecărui pixel (fără a lua în considerare locul în care este poziționat). Operațiile de vecinătate iau in considerare valorile care inconjoară pixelul nostru de interes. </p>

<p>
Fie <b>f (l,c)</b>  imaginea inițială de intrare, <b>g (l, c)</b> să fie imaginea de ieșire procesată și <b>T</b> să fie un operator pe f, definit pentru o vecinătate de pixeli <b>(l, c)</b>. Noua valoare a oricărui pixel din imaginea procesată este obținută prin combinarea unor valori ale pixelilor din imaginea inițială, plasată în jurul vecinătatii pixelului procesat. Centrul ferestrei este mutat de la pixel la pixel pana se cuprinde toata imaginea. Operatorul T este o funcție care combină valorile extrase din imagine. <b>T</b> se aplică în fiecare locație <b>(l, c)</b> pentru a obține rezultatul <b>g (l, c)</b>, în acea locație. Procesul utilizează numai pixelii cuprinsi in vecinatatea selectata. In <b>Figura 1</b> poate fi vizualizat procesul pentru o vecinatate de 3x3.</p>

<h2 id="2"> 2. Filtrarea liniara </h2>

<p> Filtrarea liniara se caracterizeaza prin combinarea liniara a valorilor aflate in vecinatartea unui pixel. Din punct de vedere matematic acest tip de filtrare se poate exprima ca: <br><br>

$$\large g(l,c) = \sum_{(m,n)\epsilon V}w_{mn}f(m+l,n+c)  $$        

<p> <b>w_{mn}</b> se numesc ponderi si sunt niste constante asociate fiecarei pozitii din fereastra de filtrare. Vecinatatea impreuna cu ponderile asociate se numeste <b>masca sau nucleu de filtrare.</b> Operatia de filtrare liniara nu este altceva decat o convolutie 2D in care fiecare intensitate din fereastra este inmultita cu coeficientul corespunzator pozitiei. <b>Figura 2.1</b> reda un exemplu de convolutie pentru o fereastra de dimensiune 3x3. </p>

<p>Operatia matematica este aplicata pentru fiecare fereastra(vecinatate) din imagine obtinandu-se intotdeauna o noua valoare care depinde de toti pixelii situati in nucleul de filtrare. <b>Figura 2.2</b> descrie procedeul de fereastra "glisanta" care se foloseste la operatia de filtrare liniara.</p>

<h2 id="3"> 3. Filtrarea de netezire si contrastare </h2>
<h3 id="3-1"> 3.1 Filtrarea de netezire (medie aritmetica) </h3>
<p> Filtrarea de netezire are ca scop principal uniformizarea regiunilor din imagine. Uniformizarea regiunilor poate reduce si aparitia zgomotului alb de distributie gaussiana. Pentru filtrul de mediere coeficientii de filtrare trebuie sa aiba aceeasi valoare, iar suma lor trebuie sa fie 1. In acest mod pentru o regiune uniforma noua valoare obtinuta dupa filtrare ar tebui sa ramana identica. <p>
  
<p>Din aceasta cauza filtrul de mediere tinde sa "blureze" imaginea. Din punct de vedere al frecventei acest tip de filtrare se comporta ca un filtru trece-jos. Trecerea intre regiuni diferite de gri nu mai este atat de vizibila. In <b>Figura 3.1.</b> se pot observa rezultatele blurarii cu un nucleu de 11x11 respectiv 7x7 pixeli. In mod evident, cu cat fereastra de filtrare efectul de "blurare" devine mai proeminent.</p>

<br>

    <img src='imagini/imag_5.png' width:150><br>

<p> In cazul in care se foloseste operatia de mediere pentru reducerea zgomotului ar trebui introduse o serie de metrici care sa masoare daca intr-adevar zgomotul a fost redus. Doua din cele mai cunoscute functii matematice folosite la aprecierea calitatii filtrarii sunt <b>MSE((Mean Squared Error) si MAE(Mean Absolute Error). Performanta se masoara pixel cu pixel intre imaginea originala(cu zgomot) si imaginea filtrata.</b>:

$$\large MSE = \frac{1}{MN}\sum_{i=0}^{M-1} \sum_{j=0}^{N-1} (f(i,j)-f_{0}(i,j))^{2}$$ 

<br>
$$\large MSE = \frac{1}{MN}\sum_{i=0}^{M-1} \sum_{j=0}^{N-1} |f(i,j)-f_{0}(i,j|$$

```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import io,util,color
import math
import time

#aplica media aritmetica pentru o vecinatate
#sunt 2 variante, una clasica si una scurta cu np.sum()
def smooth(fereastra, window_size):
    
    #PARAMETERS : fereastra,dim ferestrei
    #RETURN  :   media aritmetica a pixelilor din fereastra
    suma=0
    
    #Varianta 1
    # se parcurge fereaastra si se calculeaza media aritmetica
    # for i in range(window_size):
    #     for j in range(window_size):
    #         suma=suma+fereastra[i,j]
    # return suma/(window_size*window_size)
    
    #Varianta 2
    # se realizeaza aceeasi operatie folsind np.sum() 
    suma=np.sum(fereastra)
    return suma/(window_size*window_size)
    
def filt_mediere_aritm(img, h, w, window_size, capat):
    
    #PARAMETERS : imaginea,dim imaginii,dim ferestrei, capat
    #RETURN  :   imag filtrata
    
    #initializez imaginea pentru filtrare cu 0
    # se pierd capat linii/coloane la fiecare margine
    img_filtrata=np.zeros((h-2*capat, w-2*capat))
    
    #parcurg imaginea, decupez si aplic functia scrisa anterior pentru fiecare fereastra
    for i in range(capat, h-capat):
        for j in range(capat, w-capat):
            fereastra=img[i-capat:i+capat+1, j-capat:j+capat+1]
            img_filtrata[i-capat, j-capat]=smooth(fereastra, window_size)
    
    #returnez imag_filtra
    return img_filtrata
    
# Citesc imag color    
img = io.imread('lena.png')
print(img.dtype)

# transform in gray imaginea si schimb gama in   [0,255]
img=color.rgb2gray(img)*255

#dimensiunea imaginii
h,w=img.shape

# afisez imaginea gray
plt.figure(),plt.imshow(img,cmap='gray',vmin=0,vmax=255),plt.title('imaginea gray')

#adaug zgomot in imagine si afizes imaginea zgomotosa
#zgomot de medie 0 dispersie sigma 
sigma=10
N=np.random.normal(0,sigma,(h,w))

#adun zgomotul in imagine si afisez
img_noise=img+N
plt.figure(),plt.imshow(img_noise,cmap='gray',vmin=0,vmax=255),plt.title('imaginea cu zgomot')

#dim ferestrei si numarul de linii si coloane care se pierd pt fiecare tip de fereastra
# incerc pe rand ferestrele 3x3, 5x5, 7x7
for window_size in [3,5,7]:
    capat=math.floor(window_size/2)

    b=time.time()

    #aplic functia de filtrare liniara scrisa mai sus
    img_filt=filt_mediere_aritm(img_noise, h, w, window_size, capat)

    e=time.time()
    print('fereastra',window_size,'x',window_size,'- timp executie: ',e-b)

    #afisez imaginea filtrata
    plt.figure(),plt.imshow(img_filt,cmap='gray',vmin=0,vmax=255),plt.title('filtrata cu fereastra '+str(window_size)+'x'+str(window_size))
```

<h3 id="4-2"> 4.2 Filtrarea de netezire (medie ponderata)  </h3>
<p> Renuntati la functia <b>smooth()</b> de la exercitiul precedent. La acest exercitiu vom schimba masca de filtrare cu una care ia valori dupa o distributie gaussiana exact ca zgomotul nostru. Pe langa acest lucru, ne dorim sa scriem o metrica de performanta pentru a masura calitatea filtrarii. Vom folosi MSE(MEan Squared Error) si vom compara performanta intre o filtrare de medie aritmetica si una de medie ponderata.  </p>
<p> <strong>Hint</strong>: In care caz credeti ca eroarea va fi mai mica si de ce? </p>

```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import io,util,color
import math
import time

def filt_mediere_aritm(img, h, w, capat, mask):
    
    #PARAMETERS : imaginea,dim imaginii, capat,masca de filtrare
    #RETURN  :   imag filtrata
    
    #initializez imaginea pentru filtrare cu 0
    img_filtrata=np.zeros((h-2*capat, w-2*capat))
    
    #parcurg imaginea, decupez fereastra  o inmultesc cu masca initializata in main si calculez rez cu np.sum()
    # nu mai impart la nr de pixeli, suma ponderilor din masca e deja 1
    for i in range(capat, h-capat):
        for j in range(capat, w-capat):
            fereastra=img[i-capat:i+capat+1, j-capat:j+capat+1]
            img_filtrata[i-capat, j-capat]=np.sum(fereastra*mask)
    
    #returnez imag_filtra
    return img_filtrata
    
def MSE(img_orig, img_prel, r, c, capat):
    
    #PARAMETERS : imaginea_orig,imag_prelucrata,dim imaginii_orig, capat
    #RETURN  :   MSE intre pixelii celor 2 imagini
    
    mse=0
    
    # parcurg imaginea si fac diferenta pixel cu pixel conform formulei de mai jos
    # 1/(r*c) suma(0,r)suma(0,c) (F-F0)^2
    # PUTETI SA VA GANDITI SI LA VARIANTA VECTORIZATA(FARA FOR) -> IN PYTHON POT SA REALIZEZ OPERATIILE MATEMATICE 
    #FARA SA PARCURG ELEMNT CU ELEMENT
    for i in range(capat,r-capat):
        for j in range(capat,c-capat):
            # imaginea prelucrata e mai mica, deci e decalata cu capat
            mse=mse+(img_orig[i,j]-img_prel[i-capat,j-capat])**2
    
    # returneaza MSE 
    # impart la nr de pixeli comparati (fara marginile pierdute)
    return mse/((r-2*capat)*(c-2*capat))
    
# Citesc imag color    
img = io.imread('lena.png')
print(img.dtype)

# transform in gray imaginea si schimb gama in   [0,255]
img=color.rgb2gray(img)*255

#dimensiunea imaginii
h,w=img.shape

# afisez imaginea gray
plt.figure(),plt.imshow(img,cmap='gray')

#adaug zgomot in imagine si afizes imaginea zgomotosa
#zgomot de medie 0 dispersie sigma 
sigma=10
N=np.random.normal(0,sigma,(h,w))

#adun zgomotul in imagine si afisez
img_noise=img+N
plt.figure(),plt.imshow(img_noise,cmap='gray')

#dim ferestrei si numarul de linii si coloane care se pierd pt fiecare tip de fereastra
window_size=3
capat=math.floor(window_size/2)

#initializeaza masca de filtrare pentru media ponderata
#[[0.075,0.124,0.075],[0.124,0.2,0.124],[0.075,0.124,0.075]]
mask_ponderat=np.array([[0.075,0.124,0.075],[0.124,0.2,0.124],[0.075,0.124,0.075]])

#pt aritmetic fereastra e tot 3x3, toti coef sunt egali cu 1/9
mask_aritmetic=np.ones((3,3))/9

b=time.time()

#aplic functia de filtrare liniara scrisa mai sus pentru cele 2 cazuri(medie arit si ponderata)
img_filt_ponderat=filt_mediere_aritm(img_noise, h, w, capat, mask_ponderat)
img_filt_aritmetic=filt_mediere_aritm(img_noise, h, w, capat, mask_aritmetic)

#MSE intre imaginea originala si imaginea zgomotoasa inainte sa aplicam filtrarea
# capat=0 pt ca in cazul imaginii originale nu am pierdut inca linii si coloane, urmeaza la filtrare
print('Mse inainte de filtrare: ',MSE(img,img_noise,h,w,0))

##MSE dupa filtrarea cu medie ponderata
# COMPLETATI VOI PARMETRII
print('Mse pentru filtrarea ponderata dupa filtrare: ',MSE(img,img_filt_ponderat,h,w,capat))

##MSE dupa filtrarea cu medie aritmetica
# COMPLETATI VOI PARMETRII
print('Mse pentru filtrarea aritmetica dupa filtrare: ',MSE(img,img_filt_aritmetic,h,w,capat))

e=time.time()
print('timp executie: ',e-b)

#afisez imaginea filtrata aritmetic si ponderat
plt.figure(),plt.imshow(img_filt_ponderat,cmap='gray')
plt.figure(),plt.imshow(img_filt_aritmetic,cmap='gray')
# plt.figure(),plt.imshow(scipy.signal.convolve2d(img_noise,mask_aritmetic,mode='valid'),cmap='gray')
```
