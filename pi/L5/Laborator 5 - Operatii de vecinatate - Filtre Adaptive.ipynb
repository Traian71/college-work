<h1 id="1"> 1. Introducere </h1>

<p> Daca in laboratorele trecute operatiile matematice aplicate pe imagini au fost punctuale sau pe vecinatate, lucrarea curenta va pune accent pe <b>operatiile integrale</b>. Operatiile integrale se refera la faptul ca noua valoare de intensitae din imaginea prelucrata va depinde de toate intensitatile initiale.
    Transformatele unitare se refera la o clasa de matrici unitare folosite in reprezentarea semnalelor/imaginilor. Proprietate principala este:     
</p>

$$ A^{-1}=A^{*T} $$

<p> In cazul 1D avem urmatoarele relatii: </p><br>

<li> Vector de intrare:  </li>
$$ u = [u(0), u(1),..., u(N −1)]^T  $$
<li>  Transformarea directa:  </li>
$$ v = Au  $$
<li> Vectorul transformat:  </li>
$$ v = [v(0),v(1),...,v(N −1)]^T  $$
<li> Transformarea inversa:  </li>
$$ u=A^{-1}v=A^{*T}v $$
    
    
 
</p>
<p>
Dat fiind faptul ca (teoretic) operatia difera de la un punct la altul, nu mai putem vorbi de o filtrare globala liniara, principiul superpozitiei nu mai este respectat .
</p>
<p> 
    Adaptarea filtrelor liniare se poate face prin doua variante:
    <ul>
        <li> Modificarea formei ferestrei de filtrare </li>
        <li> Modificarea coeficientilor unei ferestre de filtrare de forma fixa </li>
    </ul>
 </p>
<h1 id="2"> 2. Filtrarea de netezire directionala adaptiva </h1>
<p>
    Termenul de fitlrare directionala se refera la forma puternic orientata a ferestrei de filtrare. Aceasta nu mai este patrata, ci va avea o forma liniara, orientata pe o anumita directie. Pentru fiecare punct al imaginii se pot considera mai multe orientari posibile pentru masca de mediere. In urma filtrarii, pentru fiecare directie se obtine o valoare, dorim ca aceasta valoare sa nu difere semnificativ de valoarea intiala a pixelului considerat. O diferenta semnificativa indica ca ne aflam pe un pixel de contur, iar pastrarea acestei noi valori ar duce la efectul de incetosare. Solutia adusa de acest filtru adaptiv este de a alege din setul de valori obtinute pe aceea care este cea mai apropiata de valoarea initiala a punctului curent. 
</p>
<p> Un posibil exemplu ar fi un filtru cu ferestre directionale de 3 pixeli, care sunt alese dupa cele patru directii principale : orizontal, vertical si cele doua diagonale, adica: <span> $$ 0^o\ ,45^o\ ,90^0\ ,135^o $$ </span>
</p>
<p>
    Pe scurt, algoritmul descris ar fi in felul urmator:
    <ol>
        <li> Pentru fiecare pixel al imaginii: </li>
        <ol type="1" style="list-style-type: decimal">
            <li> Calculam mediile pe cele 4 directii a pixelilor din jurul pixelului curent </li>
            <li> Dintre aceste 4 medii, alegem valoarea care este cea mai apropiata de valoarea pixelului curent </li>
            <li> Aceasta valoare devine valoarea de iesire a filtrului pentru  pixelul curent </li>
        </ol>
    </ol>
</p>
<h1 id="3"> 3. Adaptarea comportamentului unei ferestre fixe</h1>
<p>
    Ideea de a accepta valoarea produsa de medierea valorilor selectate de fereastra filtrului numai daca acesta valoare este suficient de apropiata de valoarea originala se poate aplica si pentru filtrarea de netezire cu fereastra si coeficienti ficsi. In acest caz rezultatul netezirii ıntr-un punct va fi acceptat doar daca diferenta fata de valoarea originala estemai mica decat un prag impus.
</p>

$$ \hat{f}(y,x) = \left\{ \begin{array}{2} 
\sum_{k,l \in W} w_kl f(y-k,x-l),\ daca |f(y,x)- \sum_{(k,l)\in W}w_kl f(y-k,n-l)| < T \\
f(y,x),\ in\ rest\\
\end{array}
\right.
$$

<h1 id="4"> 4. Adaptarea coeficientilor - LLMMSE (Lee) </h1>
<p> Aceasta varianta de filtrare linaira adaptiva are ca principiu recalcularea coeficientilor mastii de filtrare in fiecare punct al imaginii. Acesta reprezinta o combinatie liniara convexa a imaginii zgomotase si a imaginii obtinute printr-o filtrare liniara de mediere </p>
$$ \hat{f} = \alpha f + (1 - \alpha)\bar{f} $$

<p> Imaginea zgomoatasa f provine dintr-o imagine corecta g, la care s-a adaugat un zgomot alb aditiv de medie nula, necorelat cu imaginea, astfel: </p>
$$ \hat{f}=\alpha(g+n)+(1-\alpha)\bar{(g+n)}=\alpha g + \alpha n + (1-\alpha)\bar{g} $$

<p> Filtrul adaptiv este definit de coeficientul &alpha;; Acesta este calculat local impunand optimizarea unei masuri de calitate obiective, eroare patratica medie intre imaginea originala g si rezultatul filtrarii ^f.</p>
$$ \bar{\epsilon^2} = \alpha^2 \sigma^2_n + (\alpha-1)^2\sigma^2_g $$
<p> Gasirea coeficientului &alpha; optim inseamna gasirea minimului erorii patratice medie, deci anularea derivatei aceteia </p>
$$ \alpha = 1 - \frac{\sigma^2_n}{\sigma^2_f} $$
<p> Expresia finala a filtrului:</p>
$$ \hat{f}=\left(1-\frac{\sigma^2_n}{\sigma^2_f}\right)f+\frac{\sigma^2_n}{\sigma^2_f}\bar{f} $$

<p> Se observa necesitatea cunoasterii puterii zgomotului si a variantei locale pentru fiecare pixel al imaginii zgomotoase. Adapatrea se face intre doua cazuri limita:
<ol>
    <li> Daca puterea de zgomot este mult mai mare decat varianta valorilor imaginii originale, atunci &alpha; = 0, si deci, filtrul optim in aceasta situatie este filtrul de mediere. </li>
    <li> Daca varianta valorilor din imagine este mult mai mare decat puterea de zgomot, deci putem presupune ca punctul curent este situat pe un contur, atunci &alpha; = 1, si deci filtrul optim este un filtru identitate ( trece tot ) </li>
   </ol>
</p>
<h1 id="5"> 5. Filtrul Bilateral </h1>
<p>
Filtrul bilateral este un filtru neliniar de reducere a zgomotului, cu propietati de pastrare a contururilor. Acesta aplica o medie ponderata a valorilor din vecinatate, unde ponderile depind atat de distanta euclidiana dintre pixelul central si cei vecini, cat si de distantele radiometrice ( diferentele de intensitate ) ale pixelilor, in asa fel incat contururile sunt pastrate.
</p>
$$ \hat{f}(y,x) = \sum_{(m,n)\in V}w_{mn}(y,x)f(y-m,x-n), w_{mn}(y,x)\in \mathbb{R} $$
$$ w_mn(y,x) = w_{pos,mn(y,x)} \times w_{val,mn}(y,x)\ /\ \sum_{(i,j)\in V} w_{pos,ij} (y,x) \times w_{val,ij} (y,x); $$

<p>
Daca presupunem ca ambele nuclee sunt functii gausiene, atunci putem nota ponderile pentru un pixel localizat la coordonatele (y,x) si un vecin localizat la coordonatele (m,n), atunci:
</p>

$$ w_{pos,mn} (y,x) = \frac{1}{ \sigma_{pos}\sqrt{2\pi} }exp\left( - \frac{m^2+n^2}{2\sigma_{pos}^2} \right) $$

$$ w_{val,mn} (y,x) = \frac{1}{ \sigma_{val}\sqrt{2\pi} } exp\left(-\frac{(f(y-m,x-n)-f(y,x))^2}{2\sigma_{val}^2}\right); $$

<code> cv.bilateralFilter() </code>
