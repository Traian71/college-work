<h2 id="1"> 1. Introducere </h2>
<p> Imbunatatirea imaginilor se refera la o clasa de metode si algoritmi prin care este marita vizibilitatea elementelor constituente din punct de vedere al sistemului vizual uman. Tinand cont de acest aspect, conceptul de "imagine imbunatatita" 
este subiectiv si difera de la persoana la persoana (aceeasi imagine poate fi perceputa diferit de 2 persoane)</p>
<p> Operatiile punctuale sunt metode care realizeaza o mapare de tip "unu la unu" intre pixelii imaginii originale si pixelii
imaginii prelucrate(imbunatatite). Altfel spus, pentru a calcula noua valoare a intensitatii unui pixel trebuie sa aplic o 
functie matematica care depinde de vechea valoare care provine din imaginea originala. Conceptul este redat sugestiv in 
<b>Figura 1</b>. Pentru pixelul <b>U</b> s-a aplicat o transformare <b>T(U) = U / 3</b> rezultand o noua valoare <b>V</b>. Aceeasi functie 
este aplicata tuturor pixelilor din partea stanga.</p> 
<br><br><br>

<figure>
    <img src='imagini/imag_1.png' width:150>
    <center><b>Figura 1. </b>Conceptul de operatie punctuala intr-o imagine</center>
</figure>
<br><br><br>

<p> Functia de transformare <b>T(U)</b> are 2 proprietati : <br><br>

<li>Noile valori obtinute trebuie sa fie cuprinse in acelasi domeniu de definitie ca si cele anterioare. Daca gama initial de valori era cuprinsa in intervalul <b>[0,M]</b> , atunci <b>T(U)</b> trebuie sa apartina acestui interval</li><br>
<li>Functia de transformare trebuie sa fie monotona (crescatoare sau descrescatoare) astfel incat ordinea pentru plaja de valori sa fie pastrata<b></b></li></p>




<h2 id="2"> 2. Modificarea contrastului </h2>
<h3 id="2-1"> 2.1 Modificarea liniara </h3>
<p> Contrastul este diferenta de luminozitate (intensitatea nivelelor de gri) intre pixelii unei regiuni. Aces lucru insemana ca el contribuie la o mai buna vizibilitate a componentelor imaginii. Una din cele mai folosite tehnici de contrastare este cea de modificare liniara a contrastului pe portiuni. Functia de transformare este definita ca: <br><br>

$$\large v=\begin{equation}
\left\{ 
  \begin{aligned}
    &\frac{T_{1}}{a}\,u,   \quad  u=[0,a) \\
    &T_{1}+\frac{T_{2}-T_{1}}{b-a}\,(u-a),   \quad  u=[a,b) &  \text{(2.1)}\\
    &T_{2}+\frac{L-1-T_{2}}{L-1-b}\,(u-b),   \quad  u=[b,L) \\
  \end{aligned}
  \right.
\end{equation}
$$


</p><br>

<p>In formula de mai sus parametrii de reglaj ai algoritmului sunt   <b>(a, b, T1, T2).</b> Dupa cum se poate observa si din structura functiei transformarea de contrast are 3 paliere (fiecare ramura a functiei). Astfel, se obtin 3 segmenete de dreapta
<b>  <font size="3"> S1  ( [0,0],[a,T1] ),  S2( [a,T1],[b,T2] ),  S3( [b,T2],[L-1,L-1] ). </font> </b> Graficul functiei poate fi urmarit in <b>Figura 2.1</b>. In functie de cele 2 perechi de parametrii (a,b) si (T1,T2) contrastul poate fi diminuat sau accentuat. Daca panta graficului este subunitara diferenta intre nivelele alaturate de gri scade (contrastul scade). In schimb daca panta devine supraunitara contrastul se mareste.</p>

<h3 id="2-2"> 2.2 Modificarea neliniara </h3>
<p> Principalul dezavantaj al tehnicilori liniare este faptul ca modificarea de contrast este acceasi pe intervale de nivele de gri. Din aceasta cauza au fost introduse metodele neliniare care permit modificarea contrastului in jurul unei regiuni mai mici de nivele de gri. Modificarea logaritmica se aplica conform ecuatiei 2.2 :  <br><br>

$$\large\begin{equation}
  \begin{aligned}
    T(u)= &\frac{L-1}{lgL}\,(u+1)   &  \text{(2.2)}\\
  \end{aligned}
\end{equation}
$$





</p>


**Raspuns la intrebarea de la 2.1:**
- **Intinderea maxima de contrast:** T1 = 0 si T2 = L-1, iar a si b sunt limitele domeniului de nivele de gri folosit de imagine (de ex. a = minimul, b = maximul). Intervalul [a, b] se intinde pe toata gama [0, L-1].
- **Binarizarea:** T1 = 0, T2 = L-1 si a = b = prag. Pixelii sub prag devin 0 (negru), iar restul devin L-1 (alb).

<p> <b>In Figura 2.2</b> se poate observa comportamentul functiei. Zonele de de gri din prima parte a gamei de valori(cele inchise) vor fi inlocuite cu tonuri de gri care sunt mai aproape de L-1. In felul acesta imaginea prelucrata va fi mult mai luminoasa 
(supraexpunere) </p>


<br><br><br>

<p>In mod evident, exista si operatia inversa pentru transformata anterioara si anume cea exponentiala. Daca in cazul precedent imaginea devenea mai luminoasa acum ea devine mai intunecata. Nivele de gri mai deschise vor fi inlocuite cu nuante mai inchise (efectul de subexpunere). In ecuatia 2.3 este redata dependenta exponentiala iar in <b>figura 2.3</b> graficul transformarii : <br><br>

$$\large\begin{equation}
  \begin{aligned}
    T(u)=&e^{\frac{u}{L-1}\,lgL}-1   &  \text{(2.3)}\\
  \end{aligned}
\end{equation}
$$

Pentru <b><font size="3">r subunitar</font></b> are acelasi comportament ca functia logaritmica, iar pentru  <b><font size="3">r supraunitar</font></b> se comporta ca o exponentiala</p><br><br>

$$\large\begin{equation}
  \begin{aligned}
    T(u)=&(L-1)\,\frac{u^{r}}{(L-1)^{r}}  &  \text{(2.4)} \\

<h2 id="3"> 3. Egalizare de histograma </h2>
<p> Desi tehnicile de contrastare prezentate anterior si-au demonstrat eficienta de-a lungul timpului, ele nu sunt automate deoarece nu reusesc sa detecteze din ce cauza vizibilitatea nu este cea mai buna intr-o imagine. Aici intervine egalizarea de histograma care este o tehnica de imbunatatire automata ce foloseste informatii referitoare la nivelele de gri. <p>

<p> Histograma caracterizeaza distributia nivelelor de gri ce apartin unei imagini. Cu alte cuvinte, in histograma se regasesc informatii despre probabilitatea de aparitie (de cate ori apare in imagine fiecare nivel de gri) a fiecarui nivel de gri in cadrul imaginii. Pentru egalizarea de histograma nu se foloseste histograma propriu-zisa, dar prin intermediul acesteia se poate calcula histograma cumulativa care face parte din functia de transformare. Histograma cumulativa reprezinta probabilitatea ca in imagine sa existe intensitati mai mici pentru valorile de gri in raport cu o valoare curenta Ecuatiile 3.1 si 3.2 impreuna cu Figura 3.1  redau matematic si grafic cele 2 tipuri de histograme. </p>


<p> 

$$
\large\begin{equation*}
h(u)=\frac{1}{M*N}\left( \sum_{m=0}^{M-1} \sum_{n=0}^{N-1} \Delta(f(m,n)-u) \right) ,u=0,1,...,L-1      (3.1)
\end{equation*}
$$


$$
\large\begin{equation*}
H(u)=\left( \sum_{i=0}^{u} h(i) \right) ,u=0,1,...,L-1      (3.2)
\end{equation*}
$$

<br><br><br>

<figure>
    <img src='imagini/imag_6.png' width=800><br>
    <center><b>Figura 3.1. </b>Histograma si histograma cumulativa asociata unei imagini</center>
</figure>
<br><br><br>
</p>


<p> Desi metoda e satisfacatoare pentru multe imagini, egalizarea histogramei nu este uniforma si nici nu ocupa tot spatiul (are aspect de pieptene). Dupa aplicarea egalizarii nivelele de gri vor fi mai putine decat in faza initiala. Acest lucru se datoreaza faptului ca egalizarea atribuie unei game foarte apropiate de nivele de gri aceeasi valoare. Harta de transformare pentru egalizare este descrisa in ecuatia 3.3. H este histograma cumulativa,iar L numarul de nivele de cuantizare.

$$\large\begin{equation}
  \begin{aligned}
    T(u)=round(\frac{H(u)-H(0)}{1-H(0)}\,(L-1))  &  \text{(3.3)} \\
  \end{aligned}
\end{equation}
$$


</p>

<h3 id="4-1"> 4.1 Modificarea liniara a contrastului </h3>
<p> Implementati algoritmul de modificare neliniara a contrastului avand in vedere urmatoarele sugestii: <br><br>

<li> Folositi o functie care aplica formula matematica pentru un pixel din imagine </li>
<li> Folositi o noua functie care aplica transformarea creata anterior pentru fiecare pixel din imagine </li>
<li> Apelati functia de la punctul 2 si afisati imaginea modificata </li>
<li> Observati rezultatete pentru <b>a=140, b=200, T1=70, T2=230</b> si pentru <b>a=20, b=220, T1=70, T2=160 </b></li>
<li> Observati rezultatete si pentru <b>intinderea maxima de contrast</b> si <b>binarizare</b>Puteti folosi orice valoare pentru parametrii cat timp respectati principiile celor 2 situatii</li>
</p>


    
 
    

```python
from skimage import io
import numpy as np
import matplotlib.pyplot as plt
import time

# functia care aplica transformarea pentru un pixel din imagine
def val_noua_pixel_curent(a, b, T1, T2, u, L):
    if u < a:
        v = T1 / a * u
    elif u < b:
        v = T1 + (T2 - T1) / (b - a) * (u - a)
    else:
        v = T2 + (L - 1 - T2) / (L - 1 - b) * (u - b)
    return v

# functia care aplica functia precedenta pentru transformarea tuturor pixelilor din imagine
def aplica_contrast_liniar(a, b, T1, T2, L, img, h, w):
    img_modif = np.zeros([h, w], dtype='uint8')
    for i in range(h):
        for j in range(w):
            img_modif[i, j] = round(val_noua_pixel_curent(a, b, T1, T2, int(img[i, j]), L))
    return img_modif


# citim imaginea
img = io.imread('lena_gray.jpeg')

# Afisam imaginea originala
plt.figure(),plt.imshow(img, cmap='gray')

# Calculati dimeniunea imaginii cu functia shape
h, w = img.shape
L = 256

begin=time.time()
# apeleaza functia aplica_contrast_liniar() pentru imaginea aleasa folosind pentru parametrii valorile din enunt
img1 = aplica_contrast_liniar(140, 200, 70, 230, L, img, h, w)

# printeaza timpul de executie al prelucrarii efective
end=time.time()
print('timp executie: ',end-begin)

# Afisarea imaginii prelucrate cu nivele de gri
# vmin/vmax fixe ca imshow sa nu reintinda singur contrastul
plt.figure(),plt.imshow(img1, cmap='gray', vmin=0, vmax=255),plt.title('a=140, b=200, T1=70, T2=230')

img2 = aplica_contrast_liniar(20, 220, 70, 160, L, img, h, w)
plt.figure(),plt.imshow(img2, cmap='gray', vmin=0, vmax=255),plt.title('a=20, b=220, T1=70, T2=160')

# intinderea maxima de contrast: T1=0, T2=L-1, a si b = minimul si maximul din imagine
img3 = aplica_contrast_liniar(int(img.min()), int(img.max()), 0, L - 1, L, img, h, w)
plt.figure(),plt.imshow(img3, cmap='gray', vmin=0, vmax=255),plt.title('intinderea maxima de contrast')

# binarizare: a=b=prag, T1=0, T2=L-1
img4 = aplica_contrast_liniar(128, 128, 0, L - 1, L, img, h, w)
plt.figure(),plt.imshow(img4, cmap='gray', vmin=0, vmax=255),plt.title('binarizare (prag 128)')
```

<h3 id="4-2"> 4.2 Modificarea liniara a contrastului (mai eficient) </h3>
<p> Scopul acestui exercitiu este obtinerea aceluiasi rezultat anterior, dar folosind o tehnica mai eficienta de prelucrare. In primul caz sunt realizate niste operatii care se repeta. Care sunt operatiile redundante? </p>
<p> <strong>Hint</strong>: Ganditi-va la intensitatile de gri dintr-o imagine si la cum sunt distribuite ele in imagine </p>

**Raspuns:** Operatia redundanta este calculul formulei 2.1 pentru fiecare pixel: imaginea are 512 x 512 = 262144 pixeli, dar doar 256 de nivele de gri posibile, deci aceeasi valoare se calculeaza de foarte multe ori. Mai eficient este sa calculam o singura data noua valoare pentru fiecare nivel de gri (harta de transformare) si apoi doar sa o aplicam imaginii.

```python
from skimage import io
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import time

# functia care calculeaza transformarea pentru toate intensitatile de gri posibile
def harta_de_transformare(a, b, T1, T2, L):
    h=np.zeros([L, 3])
    # noua valoare pentru fiecare nivel de gri, cu functia de la 4.1 (ecuatia 2.1); R=G=B
    for i in range(L):
        h[i, :] = val_noua_pixel_curent(a, b, T1, T2, i, L)
    # normare in [0,1]
    return h / (L - 1)

# citim imaginea
img = io.imread('lena_gray.jpeg')

# Afisam imaginea originala
plt.figure(),plt.imshow(img, cmap='gray')

# Calculati dimeniunea imaginii cu functia shape
h, w = img.shape
L = 256

begin=time.time()
# apeleaza functia aharta_de_transformare() pentru fiecare nivel de gri posibil din imagine
#Folositi pentru parametrii valorile unui caz propus la exercitiul 4.1
harta = harta_de_transformare(140, 200, 70, 230, L)

# transformare hartii din array in Colormap(harta de culoare)
harta = ListedColormap(harta)

# Afisarea imaginii prelucrate cu nivele de gri folosind harta creata anterior
plt.figure(),plt.imshow(img, cmap=harta, vmin=0, vmax=255),plt.title('a=140, b=200, T1=70, T2=230 (cu harta)')

# printeaza timpul de executie al prelucrarii efective
end=time.time()
print('timp executie: ',end-begin)
```

<h3 id="4-3"> 4.3 Modificarea neliniara a contrastului </h3>
    <p> La acest exercitiu folositi transformata neliniara descrisa de functia putere din ecuatia 2.4. Folositi acelasi cod ca la exercitiul 4.2, tinand cont ca singura diferenta este formula de mapare. Pentru parametrul r incercati valorile <b>r=4</b> si <b>r=0.25</b> si observati ce se intampla </p>
    

```python
from skimage import io
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import time

# functia care calculeaza transformarea pentru toate intensitatile de gri posibile
def harta_de_transformare(r, L):
    h=np.zeros([L, 3])
    # functia putere, ecuatia 2.4; R=G=B
    for i in range(L):
        h[i, :] = (L - 1) * i**r / (L - 1)**r
    # normare in [0,1]
    return h / (L - 1)
```

```python
# citim imaginea
img = io.imread('lena_gray.jpeg')

# Afisam imaginea originala
plt.figure(),plt.imshow(img, cmap='gray')

# Calculati dimeniunea imaginii cu functia shape
h, w = img.shape
L = 256

begin=time.time()
# apeleaza functia harta_de_transformare() pentru fiecare nivel de gri posibil din imagine
#Folositi pentru parametrii valorile din enunt
harta4 = harta_de_transformare(4, L)
harta025 = harta_de_transformare(0.25, L)

# transformare hartii din array in Colormap(harta de culoare)
harta4 = ListedColormap(harta4)
harta025 = ListedColormap(harta025)

# Afisarea imaginii prelucrate cu nivele de gri folosind harta creata anterior
# r=4 -> imagine mai intunecata (ca exponentiala), r=0.25 -> imagine mai luminoasa (ca logaritmica)
plt.figure(),plt.imshow(img, cmap=harta4, vmin=0, vmax=255),plt.title('r = 4')
plt.figure(),plt.imshow(img, cmap=harta025, vmin=0, vmax=255),plt.title('r = 0.25')

# printeaza timpul de executie al prelucrarii efective
end=time.time()
print('timp executie: ',end-begin)
```

<h3 id="4-4"> 4.4 Egalizare de histograma </h3>
<p> Implementati algoritmul de egalizare de histograma folosind urmatorii pasi: 

   <li> Folositi o functie care calculeaza histograma unei imagini </li>
   <li> Folositi o functie care calculeaza histograma cumulativa a unei imagini </li>
   <li> Folositi o functie care calculeaza harta de transformare pentru egalizarea de histograma </li>
   <li> Folositi o functie care care aplica egalizarea pentru toti pixelii din imagine </li>
</p>

```python
from skimage import io
import numpy as np
import matplotlib.pyplot as plt

# fct care calculeaza histograma unei imagini
# cu principiul de la dictionare chiar daca nu folosim explicit dictionare
def histograma(img, h, w):
    his = np.zeros(256)
    # intensitatea pixelului e chiar pozitia din histograma pe care o crestem
    for i in range(h):
        for j in range(w):
            his[img[i, j]] += 1
    # histograma normata la dimensiunea imaginii
    return his / (h * w)

#fct pt histograma cumulativa normata
def hist_cum(his):
    H=np.zeros(256)
    # pe pozitia i: suma histogramei pentru nivelele 0..i
    for i in range(256):
        H[i] = his[0:i + 1].sum()
    # his e deja normata, deci si H e normata
    return H
```

```python
# calculez harta pentru egalizare conform formulei 3.3
def harta_egalizare_hist(H, L):
    harta=np.zeros(256)
    for i in range(256):
        harta[i] = round((H[i] - H[0]) / (1 - H[0]) * (L - 1))
    return harta

def aplica_egalizarea(harta, img, h, w):
    img_modif=np.zeros([h,w],dtype='uint8')
    # fiecare pixel primeste valoarea din harta de pe pozitia data de intensitatea lui
    for i in range(h):
        for j in range(w):
            img_modif[i, j] = harta[img[i, j]]
    return img_modif
```

```python
# citim imaginea
img = io.imread('lena_gray.jpeg')

# Afisam imaginea originala
plt.figure(),plt.imshow(img, cmap='gray')

# calculam si afisam hist si hist_cum
h,w=img.shape

# calculeaza histograma folosind fct histograma()
his=histograma(img, h, w)
# calculeaza histograma cumulativa folosind fct hist_cum()
H=hist_cum(his)

#afiseaza histograma si histograma cumulativa
niv=np.array(list(range(256)))
plt.figure(),plt.bar(niv,his)
plt.figure(),plt.bar(niv,H)

# calculeaza harta egalizarii si o aplica pe toata imaginea
harta=harta_egalizare_hist(H, 256)
img_modif=aplica_egalizarea(harta, img, h, w)

# Afisarea imaginea egalizata 
plt.figure(),plt.imshow(img_modif,cmap='gray')

#calculeaza histograma pentru imaginea modificata
his_img_modif=histograma(img_modif, h, w)

#afiseaza histograma pentru imaginea modificata
plt.figure(),plt.bar(niv,his_img_modif)
```
